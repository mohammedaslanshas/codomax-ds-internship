```python
import csv

with open("students.csv","w",newline="") as f:
  writer = csv.writer(f)
  writer.writerow(["name","score"])
  writer.writerow(["John",90])
  writer.writerow(["Jane",85])
  writer.writerow(["Bob",78])
  writer.writerow(["Alice",92])

with open("students.csv", "r") as f:
  reader = csv.reader(f)
  for row in reader:
    print(row)
```

```python
import csv

with open("students.csv","r",newline="") as f:
  reader =  csv.reader(f)

  header = next(reader)

  for row in reader:
    if not row:
      continue

    name = row[0]
    score = int(row[1])

    if score > 40:
      print(f"Name: {name:10} | Score: {score}")
```
