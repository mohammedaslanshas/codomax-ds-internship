```python
#function
def calculate_bmi(weight_kg, height_m):
    bmi = weight_kg / (height_m**2)
    return bmi
print(calculate_bmi(65,1.73))
```

```python
#function
def classify_bmi(bmi,overweight_threshold=25):
  if bmi >= overweight_threshold:
      return "Overweight"
  return "Normal"
print(classify_bmi(27))
print(classify_bmi(27, overweight_threshold=30))
```

```python
#function
def calculate_error_rate(correct,total):
  incorrect = total - correct
  error_rate = (incorrect / total) * 100
  return error_rate
print(calculate_error_rate(80, 100))
```

```python
#Dictionaries
students = [
    {"name": "Aslansha", "score": 78},
    {"name": "Priya", "score": 45},
    {"name": "Ravi", "score": 92}
]
for s in students:
    status = "Pass" if s["score"] >= 40 else "Fail"
    print(f"{s['name']}: {status}")
```

```python
#Dictionaries
students = [
    {"name": "Aslansha", "score": 78},
    {"name": "Priya", "score": 45},
    {"name": "Ravi", "score": 92}
]
total_score = 0
for student in students:
    total_score += student["score"]
average_score = total_score / len(students)

print(f"Average Score: {average_score:.2f}")
```

```python
students = [
    {"name": "Aslansha", "score": 74},
    {"name": "Priya", "score": 38},
    {"name": "Rohan", "score": 85},
    {"name": "Sara", "score": 40},
    {"name": "Amit", "score": 74}
]
def get_pass_list(students):
    return [s["name"] for s in students if s["score"] >= 40]

passed_students = get_pass_list(students)
print("Passed students:", passed_students)

unique_scores = {s["score"] for s in students}
print("Unique scores:", unique_scores)
print("Number of unique scores:", len(unique_scores))
```
