```python
ages = [5, 34, 67, 12, 45, 89, 23]

for age in ages:
    if age < 18:
        category = "child"
    elif age < 60:
        category = "adult"
    else:
        category = "senior"
    print(f"Age {age} → {category}")
```

```python
epoch = 1
max_epochs = 5

while epoch <= max_epochs:
    print(f"Training epoch {epoch}...")
    epoch += 1
```

```python
for age in ages:
    if age > 80:
        print("Found an outlier, stopping scan.")
        break
    if age < 10:
        continue  # skip very young ages
    print(f"Processing age: {age}")
```

```python
scores = [75, 38, 92, 40, -1, 88, 20]
pass_count = 0

print("--- Exam Score Processing ---")

for score in scores:
    if score == -1:
        print("\n[ERROR] Data entry error (-1) encountered! Stopping process immediately.")
        break
    if score >= 40:
        status = "Pass"
        pass_count += 1
    else:
        status = "Fail"

    print(f"Score: {score:3d} -> Status: {status}")

print(f"Total students passed: {pass_count}")
```
