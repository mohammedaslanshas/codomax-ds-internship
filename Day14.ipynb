```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error , r2_score

data = {
    "area_sqft": [650, 800 ,900 ,1000 ,1200 ,1350 , 1500 ,1700 ,1850, 2000 , 2200 ,2400],
    "bedrooms":  [1, 2, 2, 2, 3, 3, 3, 4, 4, 4, 5, 5],
    "age_years": [15, 10, 8, 20, 5, 12, 3, 7, 25, 2, 9, 1],
    "price_lakhs": [22, 30, 34, 32, 45, 48, 58, 65, 62, 78, 88, 95]
}
df = pd.DataFrame(data)
print(df)
```

```python
X = df[["area_sqft","bedrooms","age_years"]]
y = df["price_lakhs"]

X_train , X_test, y_train, y_test = train_test_split(X,y, test_size= 0.25, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Predictions:",y_pred)
print("Actual:" , y_test.values)
print("R2:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
```

```python
for feature, coef in zip(X.columns, model.coef_):
    print(f"{feature}: {coef:.3f}")
print("Intercept:", model.intercept_)
```

```python
new_house =[[1600,3,6]]
predicted_price = model.predict(new_house)
print("Predicted price (lakhs):", predicted_price)
```
