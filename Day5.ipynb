```python
data ={
    "name"   : ["Aslansha", "Priya" ,"Rohan" , "Sara", "Amit"],
    "score"  : [74,38,85,40,74],
    "subject": ["Math","Math","Science","Science","Math"]
}
df=pd.DataFrame(data)

print(df["score"],"\n")
print(df[["name","score"]],"\n")

print(df.iloc[0],"\n")
print(df.loc[0],"\n")

passed_df=[df["score"]>=40]
print(passed_df,"\n")

math_passed = df[(df["score"] >=40) & (df["subject"] == "Math")]
print(math_passed)
```

```python
import pandas as pd
import numpy as np
data = {
    "name": ["Aslansha", "Priya", "Rohan", "Sara", "Amit"],
    "score": [74, 38, 85, 40, 74],
    "subject": ["Math", "Math", "Science", "Science", "Math"]
}
df = pd.DataFrame(data)

df["status"] = np.where(df["score"] >= 40 , "Pass" , "Fail")
print(df)
print()

avg_scores = df.groupby("subject")["score"].mean()

print(avg_scores)
print()

failed_students = df[df["status"] == "Fail"].sort_values(by="score",ascending=True)
print(failed_students)

```
